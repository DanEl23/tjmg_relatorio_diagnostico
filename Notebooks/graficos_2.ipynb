```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- 1. DADOS (Os mesmos já extraídos) ---
raw_gov = [
    ('TJPB', 661), ('TJCE', 658), ('TJRO', 654), ('TJSC', 650), ('TJAM', 648),
    ('TJDFT', 642), ('TJPI', 638), ('TJPA', 635), ('TJMS', 632), ('TJTO', 631),
    ('TJMT', 622), ('TJPE', 621), ('TJRS', 621), ('TJAP', 620), ('TJSE', 599),
    ('TJRN', 596), ('TJPR', 595), ('TJRJ', 580), ('TJBA', 536), ('TJMG', 511),
    ('TJAL', 510), ('TJSP', 496)
]

raw_prod = [
    ('TJAP', 672), ('TJDFT', 643), ('TJAM', 635), ('TJMA', 548), ('TJPA', 548),
    ('TJSC', 540), ('TJSE', 540), ('TJTO', 492), ('TJPI', 475), ('TJMS', 470),
    ('TJPE', 457), ('TJPR', 448), ('TJRS', 422), ('TJRN', 395), ('TJES', 393),
    ('TJPB', 390), ('TJAL', 358), ('TJCE', 340), ('TJRJ', 288), ('TJMG', 270),
    ('TJBA', 260), ('TJSP', 230)
]

raw_transp = [
    ('TJCE', 120), ('TJDFT', 120), ('TJES', 120), ('TJGO', 120), ('TJMS', 120),
    ('TJPI', 120), ('TJPR', 120), ('TJRJ', 120), ('TJRN', 120), ('TJRO', 120),
    ('TJRS', 120), ('TJSC', 120), ('TJSE', 120), ('TJPB', 100), ('TJAC', 40),
    ('TJAL', 40), ('TJAP', 40), ('TJMG', 40), ('TJRR', 40), ('TJSP', 40),
    ('TJTO', 40), ('TJBA', 20)
]

raw_tec = [
    ('TJPA', 594), ('TJRR', 594), ('TJSE', 589), ('TJMT', 579), ('TJPI', 574),
    ('TJSC', 574), ('TJAM', 569), ('TJGO', 569), ('TJES', 559), ('TJMS', 549),
    ('TJPB', 549), ('TJRN', 539), ('TJAP', 534), ('TJCE', 534), ('TJAC', 524),
    ('TJTO', 494), ('TJBA', 469), ('TJRS', 424), ('TJMG', 374), ('TJRJ', 367),
    ('TJAL', 277), ('TJSP', 217)
]

# --- 2. CONFIGURAÇÃO DE ESTILO ---
sns.set_theme(style="white")
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.left'] = False # Limpeza total
plt.rcParams['ytick.left'] = False

# Lista de configurações para o loop: (Dados, Título, Cor, Nome do Arquivo)
datasets = [
    (raw_gov, 'Ranking de Governança', '#4e79a7', 'grafico_governanca.png'),
    (raw_prod, 'Ranking de Produtividade', '#f28e2b', 'grafico_produtividade.png'),
    (raw_transp, 'Ranking de Transparência', '#e15759', 'grafico_transparencia.png'),
    (raw_tec, 'Ranking de Dados e Tecnologia', '#76b7b2', 'grafico_tecnologia.png')
]

# --- 3. GERAÇÃO INDIVIDUAL ---
for data, title, color, filename in datasets:
    
    # Prepara DataFrame
    df_temp = pd.DataFrame(data, columns=['Tribunal', 'Pontos'])
    df_temp = df_temp.sort_values('Pontos', ascending=False)

    # Cria figura isolada (tamanho otimizado para slide único)
    plt.figure(figsize=(12, 8)) 
    
    # Plota
    ax = sns.barplot(
        data=df_temp,
        x='Pontos',
        y='Tribunal',
        color=color
    )
    
    # Ajustes Visuais
    plt.title(title, fontsize=20, fontweight='bold', pad=20, loc='left')
    plt.xlabel('')
    plt.ylabel('')
    plt.xticks([]) # Remove eixo X numérico
    plt.yticks(fontsize=12) # Aumenta fonte dos nomes dos tribunais
    
    # Adicionar Valores
    for p in ax.patches:
        width = p.get_width()
        plt.text(
            width + (width * 0.01), # Espaçamento leve
            p.get_y() + p.get_height() / 2,
            f'{int(width)}',
            ha='left', 
            va='center',
            fontsize=12,
            fontweight='bold',
            color='#333333'
        )
    
    # Salva e Fecha
    # bbox_inches='tight' garante que o título ou nomes longos não sejam cortados
    plt.tight_layout()
    plt.savefig(filename, dpi=300, bbox_inches='tight') 
    plt.close() # Importante para não sobrepor gráficos na memória

print("Gráficos gerados com sucesso!")
```
