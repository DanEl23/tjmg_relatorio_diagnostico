```python
# 1. Renomear colunas para coincidir com o df1
dados_tjmg_2022 = dados_tjmg_2022.rename(columns={'ano': 'Ano da Meta', 'resultado': 'Valor Apurado'})

# 2. Limpar a coluna 'Valor Apurado' (remover %, trocar vírgula por ponto)
dados_tjmg_2022['Valor Apurado'] = (
    dados_tjmg_2022['Valor Apurado']
    .astype(str)
    .str.replace('%', '', regex=False)
    .str.replace(',', '.', regex=False)
    .astype(float)
)
```

```python
# Criar um mapeamento das colunas fixas a partir do df1
# Pegamos a primeira ocorrência de cada Meta para servir de referência
referencia = final_copy.groupby('Meta')[['META_ID', 'Chave', 'Resumo']].first().reset_index()

# Mesclar essas informações no df2
dados_tjmg_2022 = dados_tjmg_2022.merge(referencia, on='Meta', how='left')

# Concatenar os dois
df_final = pd.concat([final_copy, dados_tjmg_2022], ignore_index=True)
df_final['meta_num'] = df_final['Meta'].str.extract('(\d+)').astype(int)
df_final = df_final[df_final['Meta'].isin(metas_ano_final)].sort_values(by=['meta_num', 'Ano da Meta'])
df_final = df_final.drop(columns=['meta_num'])
df_final
```

```python
df_textos_meta_final = df_final[df_final['Ano da Meta'] == 2025][['Meta', 'Nº_Meta', 'Ano da Meta', 'Valor da Meta']]
df_textos_meta_final
```

```python
df_textos_meta_final_tabela = df_final[['Meta', 'Valor Apurado', 'Ano da Meta']]
df_textos_meta_final_tabela

# 1. Garantir conversão numérica
df_textos_meta_final_tabela['Valor Apurado'] = (
    df_textos_meta_final_tabela['Valor Apurado']
    .astype(str)
    .str.replace(',', '.')
    .astype(float)
)

# 2. Usar pivot_table em vez de pivot
# aggfunc='first' pega o primeiro valor se houver duplicatas. 
# Se preferir a média, use aggfunc='mean'.
df_pivot = df_textos_meta_final_tabela.pivot_table(
    index='Meta', 
    columns='Ano da Meta', 
    values='Valor Apurado', 
    aggfunc='first' 
)

# 3. Ajustar colunas e índice
df_pivot.columns.name = None
df_pivot = df_pivot.reset_index()
df_pivot.fillna(0, inplace=True)  # Substituir NaN por 0, se necessário
df_pivot['meta_num'] = df_pivot['Meta'].str.extract('(\d+)').astype(int)
df_pivot = df_pivot[df_pivot['Meta'].isin(metas_ano_final)].sort_values(by=['meta_num'])
df_pivot = df_pivot.drop(columns=['meta_num'])
df_pivot.reset_index(drop=True, inplace=True)
df_pivot
```
